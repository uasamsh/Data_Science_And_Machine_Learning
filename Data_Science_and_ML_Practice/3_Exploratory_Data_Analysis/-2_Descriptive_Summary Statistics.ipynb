### Descriptive & Summary Statistics
* **Definition:** Descriptive statistics provide a quantitative summary of a dataset, describing a variable's values and the attributes it represents (such as **Mean**, **Max**, and **Standard Deviation**).
* **Real-Time Health Monitoring Example:** 
  * Monitoring devices continuously collect patient health data in small groups (micro-batches).
  * Summary statistics (**mean**, **max**, **standard deviation**) are calculated on incoming data to track normal ranges.
  * When unusually high or low values (anomalies) appear, the system generates **automatic alerts** to detect potential health risks early.

#### Two Categories of Descriptive Statistics
1. **Describe Observations in a Variable** (Values & Center):
   * **Sum:** Total of all values.
   * **Mean:** Average value.
   * **Median:** Middle value when data is sorted.
   * **Max:** Highest value.
2. **Describe Variable Spread** (Distribution & Variation):
   * **Standard Deviation:** How far values spread out from the mean.
   * **Variance:** Measure of data spread/variation.
   * **Counts:** Total number of observations.
   * **Quartiles:** Values that divide the dataset into four equal parts (25%, 50%, 75%).

   #### Uses for Descriptive Statistics
* **Detecting outliers:** Identifying unusually high or low values (anomalies/errors) in the dataset.
* **Planning data preparation requirements for data preprocessing:** Understanding missing values, data ranges, and spread to decide how to clean and prepare the data.
* **Selecting features for use in machine learning:** Choosing the most useful variables (columns) with meaningful variation to train machine learning models.

### Looking at summary statistics that decribe a variable's numeric values

```python
# sum of all numeric columns in the cars DataFrame
cars.sum()
```

```python
# Calculate the sum across each row (axis=1) using only numeric columns
cars.sum(axis=1, numeric_only=True)
```

```python
# Calculate the median (middle value) of each numeric column
# Formula: Sort values smallest to largest -> Middle value (or average of 2 middle values if count is even)
cars.median(numeric_only=True)
```

```python
# Calculate the mean (average) of each numeric column
# Formula: Mean = (Sum of all values) / (Total number of values)
cars.mean(numeric_only=True)
```

```python
# Find the maximum (highest) value in each column of the cars DataFrame
# Formula: Max = Largest value in the column (x_max)
cars.max()
```

```python
# Select the 'mpg' column from the cars DataFrame
mpg = cars.mpg

# Find the just row index number using idxmax (position) of the highest value in the 'mpg' column
mpg.idxmax()
```

### Looking at summary statistics that describe variable distribution

### Summary Statistics: Variable Distribution (Spread)
* **Standard Deviation (`std`):** Measures how spread out the values in a variable are from the mean (average).
  * **Low `std`:** Data points are clustered closely around the mean.
  * **High `std`:** Data points are spread out over a wide range.

```python
# Calculate the standard deviation (spread) of each numeric column
# Formula: std = sqrt( sum((x - mean)^2) / (n - 1) )
cars.std(numeric_only=True)
```

* **Variance (`var`):** Measures the average squared distance of each value from the mean, showing how spread out the data is ($\text{Variance} = \text{std}^2$).
  * **Low `var`:** Values are close to the mean.
  * **High `var`:** Values are widely spread out.

```python
# Calculate the variance (spread) of each numeric column
# Formula: var = sum((x - mean)^2) / (n - 1)  [or std^2]
cars.var(numeric_only=True)
```
