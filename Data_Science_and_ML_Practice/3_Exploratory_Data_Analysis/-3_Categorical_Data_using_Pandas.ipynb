## Summarizing categorical data using pandas

### Sentiment Analysis & Categorical Variables

#### 1. Sentiment Analysis
* **Definition:** Determining whether text data (such as public product reviews) expresses a **positive (good)** or **negative (bad)** opinion.
* **Simple Approach:**
  1. Collect public product reviews using **web scraping**.
  2. Classify specific words in the text as **positive** or **negative** to score customer feedback.

#### 2. Categorical Variable
* **Definition:** A variable that accepts only a **limited and fixed number of values** (categories), assigning each observation to a specific subgroup.
* **Example:** A `Fruits` column containing `Apple` and `Orange` divides the data into two fixed subgroups (`Apple` and `Orange`), just like classifying reviews into `Positive` or `Negative`.

#### 3. Basic 3 Ways to Describe Categorical Variables
1. **Counts:** Counting how many times each category appears (e.g., `value_counts()`).
2. **Variable Description:** Summarizing total entries, unique categories, and the most frequent value (e.g., `describe()`).
3. **Grouping:** Organizing observations into specific subgroups based on their category (e.g., `groupby()`).

```python
import numpy as np
import pandas as pd
```

### The basics

```python
# Load the mtcars dataset from the CSV file
cars = pd.read_csv('../../data/mtcars.csv')

# Rename the columns of the DataFrame
cars.columns = ['car_names','mpg','cyl','disp', 'hp', 'drat', 'wt', 'qsec', 'vs', 'am', 'gear', 'carb']

# Set the 'car_names' column as the row index
cars.index = cars.car_names

# Display the first 10 rows
cars.head(10)
```

```python
# Count how many cars have each unique number of carburetors ('carb' column)
# Output meaning: Left side = number of carburetors (4, 2, 1, 3, 6, 8) | Right side = count of cars (e.g., 4 -> 10 means 10 cars have 4 carburetors)
cars['carb'].value_counts()
```

```python
# Select only the 5 categorical columns from 'cars' and store them in a new DataFrame 'cars_cat'
# (Note: This prepares the subset of categorical columns before applying groupby)
cars_cat = cars[['cyl', 'vs', 'am', 'gear', 'carb']]

# Display the first 5 rows of the new 'cars_cat' DataFrame
cars_cat.head()
```

```python
# Generate statistical description of the 'cars_cat' DataFrame to understand the distribution of categorical variables
# Group the 'cars_cat' DataFrame into subgroups based on the 'gear' column (3-gear, 4-gear, and 5-gear cars)
gears_group = cars_cat.groupby('gear')

# Generate summary statistics (count, mean, std, min, max, etc.) separately for each gear group to compare them
gears_group.describe()
```

### Transforming variables to categorical data type

```python
# Create a new column 'group' in the 'cars' DataFrame by converting 'gear' into a 'category' data type
cars['group'] = pd.Series(cars.gear, dtype='category')
```

```python
# Check the data type of the 'group' column
# Output meaning:
# - CategoricalDtype: Column is now stored as categories instead of regular numbers
# - categories=[3, 4, 5]: Only these 3 fixed groups exist in this column
# - ordered=False: Categories are treated as equal groups without a specific ranking order
# - categories_dtype=int64: The category labels (3, 4, 5) are written as integers

cars['group'].dtypes # Verify the data type of the new 'group' column (confirms it is Categorical with categories [3, 4, 5])
```

```python
cars['group'].value_counts() # Count how many cars belong to each gear group (3, 4, or 5 gears)
```

### Describing categorical data with crosstabs

```python
# Create a cross-tabulation table to count combinations of 'am' (0=Automatic, 1=Manual) and 'gear' (3, 4, 5)
# Output meaning:
# - Row 0 (Automatic cars): 15 cars have 3 gears, 4 cars have 4 gears, 0 cars have 5 gears
# - Row 1 (Manual cars): 0 cars have 3 gears, 8 cars have 4 gears, 5 cars have 5 gears
pd.crosstab(cars['am'], cars['gear'])
```
