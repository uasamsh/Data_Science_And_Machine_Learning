### Definitions (NumPy Basics)
* **Array:** A one-dimensional container for elements that are all of the same data type.
* **Matrix:** A two-dimensional container for elements that are stored in an array.

## Using NumPy to perform arithmetic operations on data
### Basic Arithmetic Operators in Python

| Operator | Arithmetic Operation |
| :--- | :--- |
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | Division |

```python
# Import NumPy library as 'np' for working with arrays and math
import numpy as np
 # Import 'randn' function to generate random numbers (standard normal distribution)
from numpy.random import randn 
```

```python
# Round printed numbers to 2 decimal places for cleaner display
np.set_printoptions(precision=2)
```

## Creating arrays
### Creating arrays using a list

```python
a = np.array([1, 2, 3, 4, 5,6]) # Create a NumPy array 'a' with elements 1 to 5
a
```

```python
b= np.array([[1, 2, 3], [4, 5, 6]]) # Create a 2D NumPy array (Matrix) 'b' with two rows and three columns
b
```

### Creating arrays via assignment

```python
# Lock the random number generator with seed 25 so the output stays the same every time
np.random.seed(25)

# Generate 6 random numbers and multiply each by 36
c = 36*np.random.randn(6)

# Display the array c
c
```

```python
d= np.arange(1,30) # Create a NumPy array 'd' with values starting from 1 to 30 (exclusive)
d
```

```python
e= np.arange(1,50,5) # Create a NumPy array 'd' with values starting from 1 to 50 (exclusive) with a step of 5
e
```

```python
a = np.array([1, 2, 3, 4, 5,6]) 
a*10
```

```python
# a output + c output

a = np.array([1, 2, 3, 4, 5,6]) 

#output c
c= np.array([  8.22,  36.97, -30.23, -21.28, -34.45,  -8.  ])
c+a
```

```python
# Subtract each element of 'a' from 'c'
c - a
```

```python
# Multiply corresponding elements of c and a
c * a
```

```python
# Divide each element of 'c' by the corresponding element of 'a'
c / a
```

### Multiplying matrices and basic linear algebra

```python
# Divide each bill amount 100, 200, 300 by the number of people 2, 4, 5
total_bill = np.array([100, 200, 300])
num_people = np.array([2, 4, 5])
total_bill / num_people
```
