## Removing duplicates
* Find and delete repeated rows in a DataFrame to keep data clean.
* Use `.duplicated()` to check which rows are duplicates (`True` or `False`).
* Use `.drop_duplicates()` to remove repeated rows or duplicates in a specific column.

```python
# Create a simple DataFrame with duplicate rows
data = {'name': ['Ali', 'Sara', 'Ali', 'Zara', 'Sara'],
        'age': [20, 22, 20, 25, 22]}

df = pd.DataFrame(data)
df
```

```python
# Check for duplicate rows (True means duplicate)
df.duplicated()
```

```python
# Remove all duplicate rows
df.drop_duplicates()
```

```python
# Remove duplicates based only on the 'name' column
df.drop_duplicates(['name'])
```

```python
# Create a DataFrame with 3 columns containing duplicate rows
DF_obj = DataFrame({'column 1': [1, 1, 2, 2, 3, 3, 3],
                    'column 2': ['a', 'a', 'b', 'b', 'c', 'c', 'c'],
                    'column 3': ['A', 'A', 'B', 'B', 'C', 'C', 'C']})

# Display the DataFrame
DF_obj
```

```python
# Check which rows are exact duplicates (returns True for duplicate rows)
DF_obj.duplicated()
```

```python
# Create a new DataFrame where row 5 has 'D' in column 3 instead of 'C'
DF_obj = DataFrame({'column 1': [1, 1, 2, 2, 3, 3, 3],
                    'column 2': ['a', 'a', 'b', 'b', 'c', 'c', 'c'],
                    'column 3': ['A', 'A', 'B', 'B', 'C', 'D', 'C']})
```

```python
# Remove duplicate rows by checking only 'column 3' (keeps the first occurrence)
DF_obj.drop_duplicates(['column 3'])
```
