```python
import numpy as np
import pandas as pd

from pandas import Series, DataFrame
```

### Concatenating data
* Combine two or more tables into one using `pd.concat()`.
* Stack rows top-to-bottom with `axis=0` or join columns side-by-side with `axis=1`.
* Use `.join()` to merge two tables side-by-side based on their index.

```python
DF_obj = DataFrame(np.arange(36).reshape(6,6))
DF_obj
```

```python
DF_obj_2 = DataFrame(np.arange(15).reshape(5,3))
DF_obj_2
```

```python
pd.concat([DF_obj, DF_obj_2], axis=1)
```

```python
pd.concat([DF_obj, DF_obj_2])
```

```python
# Create two simple DataFrames with matching index
df1 = pd.DataFrame({'name': ['Ali', 'Sara']}, index=[0, 1])
df2 = pd.DataFrame({'city': ['Lahore', 'Karachi']}, index=[0, 1])

# Join two DataFrames side-by-side based on index
df1.join(df2)
```

```python
# Create two simple DataFrames
df1 = pd.DataFrame({'name': ['Ali', 'Sara'], 'age': [20, 22]})
df2 = pd.DataFrame({'name': ['Zara', 'Omar'], 'age': [25, 24]})

# Join tables top-to-bottom (rows)
df = pd.concat([df1, df2], ignore_index=True)
df
```

```python
# Join tables side-by-side (columns)
pd.concat([df1, df2], axis=1)
```

## Sorting data
* Arrange DataFrame rows in a specific order using `.sort_values()`.
* Use `ascending=True` to sort from smallest to largest (A to Z).
* Use `ascending=False` to sort from largest to smallest (Z to A).

```python
# Sort the DataFrame rows in descending order (largest to smallest) based on column index 5
DF_sorted = DF_obj.sort_values(by=[5], ascending=[False])

# Display the sorted DataFrame
DF_sorted
```

```python
# Sort DataFrame by 'age' from smallest to largest
df.sort_values(by='age', ascending=True)
```

```python
# Sort DataFrame by 'age' from largest to smallest
df.sort_values(by='age', ascending=False)
```

## Merging data
* Combine two DataFrames based on a common column using `pd.merge()`.
* Use the `on` parameter to specify the matching column (e.g., `id`).
* Use the `how` parameter (`inner`, `left`, `right`, `outer`) to control which rows to keep.

```python
# Create two simple DataFrames with a common column 'id'
df_names = pd.DataFrame({'id': [1, 2, 3], 'name': ['Ali', 'Sara', 'Zara']})
df_city = pd.DataFrame({'id': [1, 2, 4], 'city': ['Lahore', 'Karachi', 'Multan']})

# Merge both tables based on the common 'id' column
pd.merge(df_names, df_city, on='id', how='inner')
```
