## Creating labels and annotations
## Labels and Annotations Overview

Adding labels and annotations to data visualizations makes your charts easier to read and helps tell a clearer story with your data.

This topic covers three main tasks:
* **Labeling plot features:** Adding titles and names to the X-axis and Y-axis.
* **Adding a legend to your plot:** Showing a key box that explains what each line, bar, or color represents.
* **Annotating your plot:** Adding custom notes and arrows directly onto specific data points in the chart.

### Labeling plot features
Adding titles and names to the X-axis and Y-axis.

## Labeling Plot Features Overview

There are two ways to add titles and axis labels in Matplotlib:
* **Functional method:** Calling functions directly on `plt` (such as `plt.xlabel()`, `plt.ylabel()`, and `plt.title()`).
* **Object-oriented method:** Calling methods directly on an `axes` object (such as `ax.set_xlabel()`, `ax.set_ylabel()`, and `ax.set_title()`).

## Functional method in Labeling Plot Features

````python
# Create a sequence of 9 numbers (1 to 9) to match the 9 values in y
x = range(1, 10)

# Create a list of 9 values for the height of each bar
y = [1, 2, 3, 4, 0.5, 4, 3, 2, 1]

# Draw the bar chart
plt.bar(x, y)

# Add labels and Title to the X-axis and Y-axis using the functional method
plt.xlabel('Range')
plt.ylabel('Values')
plt.title('Range and Values Bar Chart')
````

````python
# Create a list of 5 numbers representing the size of each pie slice
z = [1, 2, 3, 4, 0.5]

# Create a list of 5 vehicle names to label each slice
veh_type = ['bicycle', 'motorbike', 'car', 'van', 'stroller']

# Draw a pie chart using z for sizes and veh_type for labels
plt.pie(z, labels=veh_type)

# Display the pie chart
plt.show()
````

## Object-oriented method in Labeling Plot Features

````python
# Create figure and axes
fig = plt.figure()
# Add a plot area (axes) to the figure: [left_margin, bottom_margin, width, height]
ax = fig.add_axes([.1, .1, 1, 1])

# Add labels and title using the object-oriented method
ax.set_xlabel('X-Axis Name')
ax.set_ylabel('Y-Axis Name')
ax.set_title('Chart Title')
````

````python


# Read the CSV file from the 'data' folder into a DataFrame named 'cars'
cars = pd.read_csv('../../data/mtcars.csv')

# Assign custom names to all 12 columns in the DataFrame
cars.columns = ['car_names', 'mpg', 'cyl', 'disp', 'hp', 'drat', 'wt', 'qsec', 'vs', 'am', 'gear', 'carb']

# Select the 'mpg' column from the DataFrame
mpg = cars.mpg

# Create a blank figure canvas with a width of 12 inches and a height of 6 inches
fig = plt.figure(figsize=(14, 6))

# Add a plot area (axes) inside the figure using square brackets
ax = fig.add_axes([.1, .1, 1, 1])

# Draw a line plot of the 'mpg' values
mpg.plot()

# Set tick positions from 0 to 31 on the X-axis for all 32 cars
ax.set_xticks(range(32))

# Set car names as X-axis tick labels, rotated 60 degrees with medium font size
ax.set_xticklabels(cars.car_names, rotation=60, fontsize='medium')

# Set the main title of the chart
ax.set_title('Miles per Gallon in mtcars Dataset')

# Set the label for the X-axis
ax.set_xlabel('Car Names')

# Set the label for the Y-axis
ax.set_ylabel('Miles per Gallon')
````

# Adding a legend to your plot
Showing a key box that explains what each line, bar, or color represents.



#### The functional method 
You pass a list of names directly inside plt.legend() and set its position using loc.

### The object-oriented method 
You call .legend() directly on the ax object, or pass label='...' inside each ax.plot() call.


#### The functional method  in legend 

````python
# Draw a pie chart using values from list z
plt.pie(z)

# Add a legend using veh_type names at the best automatic location
plt.legend(veh_type, loc='best')

# Display the chart
plt.show()
````

````python
# Make the chart wider (8 inches wide, 5 inches tall) so there is room on the side
plt.figure(figsize=(8, 5))

# Draw the pie chart
plt.pie(z)

# Place the legend outside on the right side of the pie chart
plt.legend(veh_type, loc='center left', bbox_to_anchor=(1, 0.5))

# Display the chart
plt.show()
````

## Changing Legend Location in Matplotlib

You can move the legend box anywhere inside the chart using the `loc` parameter inside `.legend()`. You can write either the **Location String** (in quotes) or its **Location Code** (number).

### Syntax
```python
plt.legend(loc='location_name')
# OR
plt.legend(loc=location_code)

### Annotating Your Plot
Adding custom notes and arrows directly onto specific data points in the chart to explain important values.

**Syntax:**

.annotate('Note Text', xy=(x_point, y_point), xytext=(x_text, y_text), arrowprops=dict(facecolor='black', shrink=0.05))

## Customizing Arrows with `arrowprops` in `.annotate()`

The `arrowprops` dictionary controls how the arrow looks when pointing from your text (`xytext`) to the data point (`xy`).

### Syntax
```python
plt.annotate('Note Text',
             xy=(x_point, y_point),
             xytext=(x_text, y_text),
             arrowprops=dict(facecolor='black', width=2, headwidth=8, frac=0.2, shrink=0.05))

````python
# Find and return the highest (maximum) value in the 'mpg' column
mpg.max()
````

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 25, 45, 30, 20]

plt.plot(x, y, marker='o')

# Annotate the peak point (3, 45) with a customized arrow
plt.annotate('Peak Value',
             xy=(3, 45),
             xytext=(3.6, 40),
             arrowprops=dict(facecolor='red', width=2, headwidth=10, shrink=0.08))

plt.show()
````

````python
# Create a blank figure canvas with a width of 12 inches and a height of 6 inches
fig = plt.figure(figsize=(14, 6))
# Add a plot area (axes) inside the figure using square brackets
ax = fig.add_axes([.1, .1, 1, 1])

# Draw a line plot of the 'mpg' values
mpg.plot()

# Set tick positions from 0 to 31 on the X-axis for all 32 cars
ax.set_xticks(range(32))

# Set car names as X-axis tick labels, rotated 60 degrees with medium font size
ax.set_xticklabels(cars.car_names, rotation=60, fontsize='medium')

# Set the main title of the chart
ax.set_title('Miles per Gallon in mtcars Dataset')

# Set the label for the X-axis
ax.set_xlabel('Car Names')
ax.set_ylabel('Miles per Gallon')

# Set the Y-axis range from 0 (bottom) to 45 (top)
ax.set_ylim([0, 45])

# Add an annotation with a black arrow pointing to the highest mpg value on the chart
ax.annotate('Maximum Value',
            xy=(mpg.idxmax(), mpg.max()),
            xytext=(mpg.idxmax() + 2, mpg.max() - 5),
            arrowprops=dict(facecolor='black', shrink=0.05))
# OR 
# Add an annotation with a black arrow pointing to 'Toyota Corolla' at (19, 33.9)
# ax.annotate('Toyota Corolla', xy=(19, 33.9), xytext=(21, 35), arrowprops=dict(facecolor='black', shrink=0.05))
 # create a legend with custom settings, located at the best position automatically
ax.legend(loc='best', fontsize='medium', title='Legend Title', title_fontsize='large', shadow=True, facecolor='lightgray', edgecolor='black')
````
