## Visualizing Time Series

A **time series** measures how a variable changes over time (for example, daily temperatures, monthly sales, or yearly stock prices). Python's built-in `datetime` module and **Pandas** make it easy to organize and plot time-based data.

### Spotting Trend and Seasonality
When looking at a time series chart, we look for two main patterns:
* **Trend:** The overall long-term direction of the data (going steadily up, going down, or staying flat).
* **Seasonality:** Regular, repeating ups and downs that happen at fixed intervals (such as every summer, every weekend, or every December).

### Four Types of Time Series Patterns
1. **Constant Time Series:** Data points stay around the same average level with no clear upward/downward direction and no repeating seasonal cycles.
2. **Trended Time Series:** Data points move steadily upward (or downward) over time, without repeating seasonal waves.
3. **Untrended Seasonal Time Series:** Data shows clear repeating up-and-down waves (seasons), while the overall average stays flat.
4. **Trended Seasonal Time Series:** Data combines both patterns—it has repeating seasonal waves while steadily climbing upward (or downward) over time.

```python
import pandas as pd

# Read the Superstore-Sales CSV file using the relative path from your current notebook folder
df = pd.read_csv('../../data/Superstore-Sales.csv', index_col='Order Date', encoding='unicode_escape', parse_dates=True)

# Display the first 5 rows of the DataFrame
df.head()
```

```python
# Plot the 'Order Quantity' column over time with a larger chart size (12 inches wide, 6 inches tall)
df['Order Quantity'].plot(figsize=(12, 6))
```

```python
# create a second DtaFrame 

# Pick 100 random rows from the DataFrame using a fixed seed (25) and sort them by date
df2 = df.sample(n=100, random_state=25, axis=0).sort_index()

# Add a label to the X-axis
plt.xlabel('Order Date')

# Add a label to the Y-axis
plt.ylabel('Order Quantity')

# Add the main title to the chart
plt.title('Superstore Sales')

# Plot only the 'Order Quantity' column from the 100-row sample DataFrame
df2['Order Quantity'].plot(figsize=(10, 5))
```
