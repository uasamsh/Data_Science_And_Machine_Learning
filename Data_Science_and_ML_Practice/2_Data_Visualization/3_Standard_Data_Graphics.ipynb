## Creating Standard Data Graphics

### Two Methods for Plot Building
* **Functional Method:** Builds plots quickly by calling direct functions (e.g., `plt.plot()`, `plt.bar()`). Best for simple, fast charts.
* **Object-Oriented Method:** Creates explicit figure and axes objects (`fig, ax = plt.subplots()`) for full control over every part of the chart.

```python
import numpy as np                         # For numerical calculations and working with arrays
from numpy.random import randn             # For generating random numbers (standard normal distribution)
import pandas as pd                        # For data manipulation and analysis
from pandas import Series, DataFrame       # For creating 1D Series and 2D DataFrame tables

import matplotlib.pyplot as plt            # For creating charts and visualizations
from matplotlib import rcParams            # For setting default chart styles, sizes, and formatting
```

## Creating a line chart from a list object
#### Plotting a line chart in matplotlib

```python
x = range(1,10)                  # Generate numbers from 1 to 9 for the x-axis
y = [1,2,3,4,0,4,3,2,1]          # Define values for the y-axis
plt.plot(x,y)                    # Plot a line chart using x and y values
plt.show()                       # Display the chart
```

## Plotting a line chart from a Pandas object

```python
# Set file path, load dataset, rename columns, and select 'mpg' (Miles Per Gallon - fuel efficiency) column
address = '/workspaces/python-for-data-science-and-machine-learning-essential-training-part-1-3006708/data/mtcars.csv'

cars = pd.read_csv(address)
cars.columns = ['car_names','mpg','cyl','disp', 'hp', 'drat', 'wt', 'qsec', 'vs', 'am', 'gear', 'carb']
mpg = cars['mpg']

plt.xlabel('Index')    # Set x-axis label
plt.ylabel('Miles Per Gallon')  # Set y-axis label
plt.title('Fuel Efficiency of Cars')  # Set chart title
mpg.plot()             # Create a line chart for the 'mpg' column
```

```python
df = cars[['cyl', 'wt', 'mpg']]     # Select cylinders, weight, and mpg columns
df.plot()                           # Plot all three columns as a multi-line chart
```

## Creating bar charts
#### Creating a bar chart from a list

```python
plt.bar(x,y)                    # Create a bar chart using x and y values
plt.xlabel('Index')    # Set x-axis label
plt.ylabel('Values')  # Set y-axis label
plt.title('Bar Chart Example')  # Set chart title
plt.show()                       # Display the chart
```

#### Creating bar charts from Pandas objects

```python
mpg.plot(kind='bar')  # Create a bar chart for the 'mpg' column
plt.show()  # Display the chart
```

```python
mpg.plot(kind='barh')  # Create a horizontal bar chart for the 'mpg' column
plt.show()  # Display the chart
```

### Creating a pie chart

```python
x = [1,2,3,4,0.5]      # Define a list of values for the pie slices
plt.pie(x)             # Create a pie chart from the values in x
plt.show()             # Display the pie chart
```

### Saving a plot

```python
plt.pie(x)  
plt.savefig('pie_chart.png')  # Save the pie chart as a PNG file
plt.show()  # Display the pie chart
```
