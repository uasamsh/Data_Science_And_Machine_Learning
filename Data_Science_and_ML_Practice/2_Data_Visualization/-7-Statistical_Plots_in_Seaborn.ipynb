## Statistical Plots in Seaborn

### Benefits of Statistical Plots
Statistical plots allow viewers to:
* Identify outliers
* Visualize distributions
* Deduce variable types
* Discover relationships and correlations between variables in a dataset

### Common Types of Statistical Plots
1. **Histograms:** Used to show variable distribution (how data values are spread across ranges).
2. **Scatterplots:** Used to show relationships between two variables.
3. **Scatterplot Matrices:** Show correlations and relationships between multiple variables at once in a grid.
4. **Boxplots:** Show variable spread (minimum, median, quartiles, maximum) and are useful for outlier detection.

```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame

from pandas.plotting import scatter_matrix

import matplotlib.pyplot as plt
from pylab import rcParams
import seaborn as sns
```

```python
# Display charts directly inside the Jupyter Notebook cell output
%matplotlib inline

# Set the default chart size to 5 inches wide and 4 inches tall
rcParams['figure.figsize'] = 5, 4

# Apply a white background with gridlines using Seaborn styling
sns.set_style('whitegrid')
```

### Eyeballing dataset distributions with histograms

```python
# Read the mtcars CSV file from the 'data' folder into a DataFrame named 'cars'
cars = pd.read_csv('../data/mtcars.csv')

# Assign custom names to all 12 columns in the DataFrame
cars.columns = ['car_names', 'mpg', 'cyl', 'disp', 'hp', 'drat', 'wt', 'qsec', 'vs', 'am', 'gear', 'carb']

# Set the 'car_names' column as the row index of the DataFrame
cars.index = cars.car_names

# Select the 'mpg' column from the cars DataFrame
mpg = cars['mpg']

# Draw a light-blue histogram of 'mpg' with black bar borders and alpha slight transparency
mpg.plot(figsize=(10, 5), kind='hist', width=0.9, alpha=0.8, edgecolor='DarkBlue')
```

```python
plt.hist(mpg)
plt.xlabel('Miles per Gallon')
plt.ylabel('Frequency')
plt.title('Distribution of Miles per Gallon')
mpg.plot(figsize=(10, 5), kind='hist', width=0.9, alpha=0.8, edgecolor='DarkBlue')
plt.show()
```

```python
# Plot a distribution histogram of the 'mpg' column using Seaborn
sns.displot(mpg)
```

###  Seeing scatterplots in action

```python
# Create a scatter plot of 'hp' vs 'mpg' with dark gray dots and dot size (s) set to 150
cars.plot(figsize=(10, 5), kind='scatter', x='hp', y='mpg', c=['darkgray'], s=150)
```

```python
# Set the chart size to 10 inches wide and 5 inches tall
plt.figure(figsize=(10, 5))

# Plot a scatter plot with a regression trend line showing the relationship between 'hp' and 'mpg'
sns.regplot(x='hp', y='mpg', data=cars, scatter=True)
```

### Generating a scatter plot matrix


```python
# Create a scatterplot matrix (pair plot) showing relationships between all numeric variables in the 'cars' DataFrame
sns.pairplot(cars)
```

```python
# Select only 4 specific columns ('mpg', 'disp', 'hp', 'wt') into a smaller DataFrame
cars_subset = cars[['mpg', 'disp', 'hp', 'wt']]

# Create a scatterplot matrix (pair plot) for just these 4 selected columns
sns.pairplot(cars_subset)

# Display the final plot cleanly
plt.show()
```

### Building boxplots


```python
# Create a larger boxplot comparing fuel efficiency ('mpg') grouped by transmission type ('am'), 'am' (transmission type: 0 = Automatic, 1 = Manual)
cars.boxplot(column='mpg', by='am', figsize=(10, 5))

# Create a larger boxplot comparing car weight ('wt') grouped by transmission type ('am')
cars.boxplot(column='wt', by='am', figsize=(10, 5))
```

```python
# Set the chart size to 10 inches wide and 5 inches tall
plt.figure(figsize=(8, 5))

# Create a colored Seaborn boxplot comparing 'mpg' across 'am' (transmission type: 0 = Automatic, 1 = Manual) using the 'hls' color palette
sns.boxplot(x='am', y='mpg', data=cars, hue='am', palette='hls')
```
