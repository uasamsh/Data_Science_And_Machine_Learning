# Plot Formatting in Matplotlib

Plot formatting helps you change the visual appearance of a chart to make the data clear and easy to read. You can customize colors, line styles, and data point markers.


```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame

import matplotlib.pyplot as plt
from pylab import rcParams

import seaborn as sb
```

```python
%matplotlib inline
rcParams['figure.figsize'] = 5, 4
sb.set_style('whitegrid')
```

# 1. Defining Plot Color (`color`)
You can change the color of a line or bar using the `color` parameter.
* **By Name:** Use standard color names like `'red'`, `'blue'`, `'green'`, `'salmon'`, or short codes like `'r'`, `'g'`, `'b'`.
* **By Code:** Use RGB or RGBA color codes for custom shades.


```python
# Set line color using a color name
plt.plot(x, y, color='salmon')
```

```python
## Creating a Bar Chart in Matplotlib

# This code creates a simple vertical bar chart using two lists of numbers.

# Create a sequence of 9 numbers from 1 to 9 for the X-axis
x = range(1, 10)

# Create a list of 9 values for the height of each bar on the Y-axis
y = [1, 2, 3, 4, 0.5, 4, 3, 2, 1]

# Draw a vertical bar chart using x and y values
plt.bar(x, y)
```

```python
# Define 9 values for X-axis (from 1 to 9) so it matches the 9 widths
x = range(1, 10)

# Define 9 values for Y-axis
y = [1, 2, 3, 4, 0.5, 4, 3, 2, 1]

# Define a list of custom widths for each individual bar (9 values)
# Set custom widths for each bar: first 3 thin (0.5), middle 3 wide (0.9), and last 3 thin (0.5) for output
wide = [0.5, 0.5, 0.5, 0.9, 0.9, 0.9, 0.5, 0.5, 0.5]

# Define the color for the bars
color = ['salmon']

# Draw the bar chart with custom widths, salmon color, and center alignment
plt.bar(x, y, width=wide, color=color, align='center')
```

# 2. Customizing line styles
 Line Styles (ls and lw)
You can change the pattern and thickness of a line.

Line Style (ls):

'-' or 'solid': A continuous straight line (default).

'--' or 'dashed': A line made of short dashes.

'-.' or 'dashdot': A line with alternating dashes and dots.

':' or 'dotted': A line made of small dots.

Line Width (lw): A number that controls how thick the line is (for example, lw=3).

```python
# Create a dashed line with a thickness of 3
plt.plot(x, y, ls='--', lw=3)
```

```python
x1 = range (1,10)
y1 = [9,8,7,6,5,4,3,2,1]
plt.plot(x,y)
plt.plot(x1,y1)
plt.show()
```

```python
# Create a sequence of numbers from 1 to 9 for the second line's X-axis
x1 = range (1,10)
# Create a descending list of numbers from 9 down to 1 for the second line's Y-axis
y1 = [9,8,7,6,5,4,3,2,1]
# Plot the first line (blue) using the previously defined x and y variables
plt.plot(x,y)
# Plot the second line (orange) on the same chart using x1 and y1
plt.plot(x1,y1)
# Display the final chart containing both lines
plt.show()
```

```python
plt.plot(x,y, linestyle = 'solid', linewidth = 5)
plt.plot(x1,y1, linestyle = '--', linewidth = 5)
```

# 3. Setting plot markers
 Marker Styles (marker, markersize, mew)
Markers highlight each individual data point on the line.

Marker Style (marker): Sets the shape of the point:

'o': Circle

'*': Star

's': Square

'+': Plus sign

'x': Cross

'^' or 'v': Triangle up or down

'D': Diamond

Marker Size (markersize or s): Controls how big or small the marker is.

Marker Edge Width (mew): Controls the thickness of the marker's border.

```python
# Add star markers with size 10 and border width 2
plt.plot(x, y, marker='*', markersize=10, mew=2)
```

```python
plt.plot(x, y, marker='1', markersize=10, mew=20)
plt.plot(x1, y1, marker='+', markersize=10, mew=15)
```
