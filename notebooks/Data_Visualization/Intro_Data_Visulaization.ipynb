## Why data visualization matters
* Turn raw numbers and data insights into clear charts and graphs.
* Help people understand trends, patterns, and key findings quickly.
* Communicate data effectively so businesses can make smart decisions.

## Three Types of Graphs
* **Line Graph:** Shows trends and changes in data over time.
* **Bar Graph:** Compares values across different categories using vertical bars.
* **Pie Chart:** Displays percentage and relative sizes of categories as parts of a whole.

# Create a simple Line Graph, Bar Graph, and Pie Chart using Pandas

* df['cyl'].value_counts().plot(kind='bar')   # Bar Graph (compares categories)
* df['cyl'].value_counts().plot(kind='pie')   # Pie Chart (shows % share)

## Three Different Data Visualization Types
1. **Data storytelling:** For presentations to organizational decision-makers.
2. **Data showcasing:** For presentations to analysts, scientists, mathematicians, and engineers.
3. **Data art:** For presentations to activists or the general public.



### 1. Data Storytelling
* **Design:** Clutter-free and highly focused so the audience gets the point easily.
* **Audience:** Non-analysts and nontechnical business managers.
* **Product types:** Static images and simple interactive dashboards.

## Selecting optimal data graphics
* **Area Charts:** Show total values and volume trends over time using shaded areas below a line.
* **Bar Charts:** Compare amounts across different categories using rectangular bars.
* **Line Charts:** Track changes and continuous trends over time.
* **Pie Charts:** Show percentages and parts of a whole in a circular chart.
* **Choropleths:** Use shaded or colored map boundaries (like states or countries) to compare data by region.
* **Point Maps:** Plot exact locations (such as cities or addresses) as dots on a map.

### 2. Data Showcasing
* **Design:** Highly contextual and open-ended so users can explore the data themselves.
* **Audience:** Analysts, quants, engineers, mathematicians, and scientists.
* **Product types:** Static images and interactive dashboards.

## Choosing Graphics for Data Showcasing
* **Area Charts:** Show total values and volume trends over time using shaded areas.
* **Bar Charts:** Compare amounts across different categories using rectangular bars.
* **Line Charts:** Track continuous changes and trends over time.
* **Histograms:** Show the frequency distribution of numerical data grouped into bins (ranges).
* **Scatterplots:** Show the relationship between two numerical variables using dots.
* **Scatterplot Matrices:** Display multiple scatterplots in a grid to compare relationships among several variables at once.
* **Choropleths:** Use colored map regions (such as states or countries) to compare data by area.
* **Point Maps:** Plot exact locations (like cities or addresses) as dots on a map.
* **Raster Maps:** Display pixel-based color gradients over a geographic map to show intensity (such as weather radar, where red means high values and blue/turquoise means low values).

### 3. Data Art
* **Design:** Attention-getting, creative, and bold to make a statement, educate, and motivate.
* **Audience:** Idealists, dreamers, artists, and social activists.
* **Product types:** Almost always static images.

## Choosing Graphics for Data Art
* **Line Charts:** Show clear trends and changes over time in a visually simple way.
* **Graph Networks:** Show connections and relationships between different people, items, or nodes.
* **Choropleths:** Use colored map regions (such as states or countries) to compare data across areas.
* **Something Weird and Artistic:** Creative, unique visuals designed to grab the general public's attention.

# Four Steps to Choosing Data Graphics
Follow these four simple steps to pick the right chart or graph for your data:

1. Make a list of the questions that your data visualization is meant to answer.
2. Is your data visualization type data storytelling, data showcasing, or data art?
3. What data graphic types are preferable for that type of data visualization?
4. Test out different types of data graphics with your data. Which graphic type displays the clearest and most obvious answers to your questions?

### How to Apply These Steps (Action Guide)
* **Step 1 (Define Goal):** Write down the exact problem or question your audience needs answered.
* **Step 2 (Know Audience):** Choose **Storytelling** for managers, **Showcasing** for analysts, or **Data Art** for the general public.
* **Step 3 (Shortlist Charts):** Pick matching charts (e.g., line charts for trends, bar charts for comparisons, or maps for locations).
* **Step 4 (Test & Pick):** Plot your data in 2–3 shortlisted charts and keep the one that is easiest to read at a glance.

## Choose Consistent Colors
* Use shades from the same color formula so visuals look harmonious and professional.
* Apply color theory schemes such as monochromatic, analogous, complementary, triadic, or tetradic.
* Use the AI color tool **Colormind** (`http://colormind.io/`) to generate matching color palettes and hex codes.

## Annotations as Context
* **What is an Annotation?** A short text note, label, or arrow placed directly on a chart to highlight a specific data point.
* **Why use it?** It gives context to the audience by explaining *why* a trend, spike, or drop happened in data storytelling and showcasing.
* Add short text notes on key data points to explain why trends happen in data storytelling and showcasing.

## Graphic Elements as Context
* **What are Graphic Elements?** Visual lines or markers (like trendlines, benchmarks, and single-value alerts) added to a chart.
* **Why use them?** They help the audience quickly see the true importance and direction of the data at a glance.

## Creating Context
* **How:** Add related metrics, trendlines, colors, annotations, or markers to the chart.
* **Why:** Gives the audience deeper insight into what is happening in the data.
* **When:** Most useful in **data showcasing**.

# Introduction to the matplotlib and Seaborn libraries

### Matplotlib vs. Seaborn
* **Seaborn:** Simple, clean syntax with built-in attractive themes for creating advanced charts with minimal code.
* **Matplotlib:** Highly customizable library with direct access to classes for fine-tuning and modifying every visual detail of charts and plots.

### Kinds of Data You Can Visualize
Matplotlib and Seaborn work with many types of data, including:
* Financial data
* Operational data
* Employee data
* Engagement and behavioral data

### Kinds of Graphs You Cannot Create
* Matplotlib and Seaborn **cannot** display network data in graphical layouts.

### Matplotlib
* Used for generating basic, straightforward plots such as bar charts, line charts, and scatter plots.

### Seaborn
* Specializes in statistical visualizations and plots complex graphics with minimal coding effort.
