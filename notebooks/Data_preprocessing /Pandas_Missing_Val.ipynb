## Working with Missing Data in Pandas

```python
import numpy as np
import pandas as pd

from pandas import DataFrame
```

### Filling missing values using fillna(), replace() and interpolate()
#### Handling Missing Values:

fillna(): Replaces NaN (missing) values with a specified value, mean, or adjacent value.

replace(): Swaps specific values (such as NaN or placeholders) with new values.

interpolate(): Estimates and fills missing values automatically based on surrounding numbers.

```python
data = {'names': ['steve', 'john', 'richard', 'sarah', 'randy', 'micheal', 'julie'],
        'age': [20, 22, 20, 21, 24, 23, 22],
        'gender': ['Male', 'Male', 'Male', 'Female', 'Male', 'Male', 'Female'],
        'rank': [2, 1, 4, 5, 3, 7, 6]}

df = pd.DataFrame(data)
# Put missing values (NaN) in rows 2, 3, 4 of column 1 (age)
df.iloc[2:5, 1] = np.nan
df
```

```python
# Put missing values (NaN) in rows 3, 4, 5 of column 3 (rank)
df.iloc[3:6, 3] = np.nan
df
```

```python
# Put missing values (NaN) in the entire row 3 (all columns)
df.iloc[3, :] = np.nan
df
```

```python
# Check for missing values (returns True where value is NaN, False otherwise)
df.isnull()
```

```python
# Check for non-missing values (returns True where value is present, False where NaN)
df.notnull()
```

```python
# Show only rows where 'age' is missing (NaN)
df[df['age'].isnull()]
df
```

```python
df.fillna(0)  # Fill missing values with 0
```

```python
# Fill missing values (NaN) with the value from the previous row above
df.ffill()
```

```python
# Fill missing values (NaN) with the value from the next row below
df.bfill()
```

```python
# Remove all rows that contain any missing values (NaN)
df.dropna()
```

```python
# Replace all missing values (NaN) with 0
df.replace(np.nan, 0)
```

```python
# Replace 'Male' with 'M' in the DataFrame
df.replace('Male', 'M')
```

```python
# Estimate and fill missing numeric values based on surrounding numbers
df['age'].interpolate()
```

```python
# Remove only rows where all columns are missing (NaN)
df.dropna(how='all')
```

```python
# Remove all columns that contain any missing values (NaN)
df.dropna(axis=1)
```

```python
# Remove all rows that contain any missing values (NaN)
df.dropna(axis=0)
```

```python
df.dropna(axis=0, subset=['age', 'rank']) # Remove rows where 'age' or 'rank' is missing (NaN)
```
