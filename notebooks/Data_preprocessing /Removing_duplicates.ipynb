## Removing duplicates
* Find and delete repeated rows in a DataFrame to keep data clean.
* Use `.duplicated()` to check which rows are duplicates (`True` or `False`).
* Use `.drop_duplicates()` to remove repeated rows or duplicates in a specific column.

```python
# Create a simple DataFrame with duplicate rows
data = {'name': ['Ali', 'Sara', 'Ali', 'Zara', 'Sara'],
        'age': [20, 22, 20, 25, 22]}

df = pd.DataFrame(data)
df
```

```python
# Check for duplicate rows (True means duplicate)
df.duplicated()
```

```python
# Remove all duplicate rows
df.drop_duplicates()
```

```python
# Remove duplicates based only on the 'name' column
df.drop_duplicates(['name'])
```
