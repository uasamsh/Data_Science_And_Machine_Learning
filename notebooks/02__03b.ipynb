# Filtering and Selecting

1. Filtering and Selecting means extracting specific rows or columns from a dataset based on conditions or labels.
2. Selecting chooses particular columns or rows by name or position, while filtering keeps only the rows that meet a certain rule (such as values greater than 50).
```[cite: 21]

````python
import numpy as np
import pandas as pd
from pandas import DataFrame
````

````python

# 1. Create numbers from 0 to 90 (with a gap of 3) and arrange into 10 rows and 3 columns
data = np.arange(0, 90, 3).reshape(10, 3)

# 2. Create a list of custom names for all 10 rows
rows = ['row 1', 'row 2', 'row 3', 'row 4', 'row 5', 'row 6', 'row 7', 'row 8', 'row 9', 'row 10']

# 3. Create a list of custom headings for all 3 columns
cols = ['column 1', 'column 2', 'column 3']

# 4. Build the DataFrame (table) using the data, row names, and column headings
df = pd.DataFrame(data, index=rows, columns=cols)

# 5. Display the final DataFrame on the screen
df
````

# Numpy Indexing And Slicing
1. Indexing means accessing specific items from data structures like arrays, Series, or DataFrames.

The `iloc` indexer stands for **integer location** and selects rows and columns strictly by their numeric position (starting from `0`).
It uses square brackets `[row_position, column_position]` to fetch exact cells, rows, or slices without needing row or column names.

2. Slicing allows us to select a range or portion of rows and columns from a dataset.
There are three main types of indexing in Python: simple indexing, Boolean indexing, and fancy indexing.


````python
# Select the value at row position 0 (first row) and column position 1 (second column)
df.iloc[0, 1]
````

### Simple Indexing
Simple indexing selects a single column, row, or specific cell directly using its label or integer position inside square brackets `[...]`.
It is the quickest way to access one specific piece of data without using conditions or lists.

````python
# Import NumPy library for working with numbers and arrays
import numpy as np

# Import Pandas library for creating and working with tables (DataFrames)
import pandas as pd

# 1. Create numbers from 0 to 90 (with a gap of 3) and arrange into 10 rows and 3 columns
data = np.arange(0, 90, 3).reshape(10, 3)

# 2. Create a list of custom names for all 10 rows
rows = ['row 1', 'row 2', 'row 3', 'row 4', 'row 5', 'row 6', 'row 7', 'row 8', 'row 9', 'row 10']

# 3. Create a list of custom headings for all 3 columns
cols = ['column 1', 'column 2', 'column 3']

# 4. Build the DataFrame (table) using the data, row names, and column headings
df = pd.DataFrame(data, index=rows, columns=cols)

# 5. Display the final DataFrame on the screen
df
````

````python
# Use iloc to locate a specific cell by row and column numbers (0-based)
df.iloc[0, 1]
````

````python
# 1. Select the value at row position 1 (second row) and column position 1 (second column)
df.iloc[1, 1]
````

````python
# 2. Change the value at row position 0 (first row) and column position 1 (second column) to 20
df.iloc[0, 1]= 20
df
````

### Fancy indexing
Fancy indexing allows us to select multiple specific columns or rows at once by passing a list of labels or positions.
It uses double square brackets `[[...]]` to extract non-consecutive items in any desired order.

````python
#  Pick only 'column 1' and 'column 3' together
df[['column 1', 'column 3']]
````

````python
# Pick only rows at position 0, 2, and 5
df.iloc[[0, 2, 5]]
````

````python
# Pick rows 0 and 4, and only their columns 0 and 2
df.iloc[[0, 4], [0, 2]]
````

````python
# Select rows at positions 1, 2, 4 (row 2, row 3, row 5) and columns at positions 1, 2 (column 2, column 3)
df.iloc[[1, 2, 4], [1, 2]]
````

## Boolean or Conditional indexing 
## Boolean values are also called a Boolean mask.
#### Boolean indexing is done through Comparison (Arithmatic) operators  (> < = <= => == !=) and Masking.
Filters data in a DataFrame using a condition that evaluates to `True` or `False`.
Only the values or rows that satisfy the condition (`True`) are displayed, while the rest are hidden.

````python
# 1. Check which values in the DataFrame are greater than 50 (returns True or False)
df > 50
````

````python
# 2. Filter and display only the values that are greater than 50
df[df > 50]
````

````python
# 3. Filter and display only the rows where 'column 1' has values greater than 30
df[df['column 1'] > 30]
````

````python
mask = df > 30
mask
````
